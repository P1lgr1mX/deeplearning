# Khái quát hóa (Generalization)
:label:`sec_generalization_basics`

Hãy tưởng tượng hai sinh viên đang chuẩn bị cho kỳ thi cuối kỳ:
* **Ellie Siêu Phàm**: Chuẩn bị bằng cách **học vẹt** toàn bộ đáp án của các đề thi năm trước. Nếu đề thi lặp lại câu hỏi cũ, Ellie sẽ đạt điểm tuyệt đối (100%). Nhưng nếu gặp câu hỏi mới, cô ấy sẽ "đóng băng".
* **Irene Quy Nạp**: Trí nhớ kém hơn nhưng có khả năng **tìm ra quy luật**. Dù đề thi có câu hỏi mới, Irene vẫn có thể duy trì phong độ đạt điểm giỏi (90%).

Là những nhà khoa học học máy, mục tiêu của chúng ta là tìm ra **quy luật** (như Irene) chứ không phải học vẹt dữ liệu (như Ellie). Khả năng áp dụng những gì đã học vào dữ liệu hoàn toàn mới chưa từng thấy chính là **Khả năng khái quát hóa (Generalization)** — vấn đề cốt lõi của Machine Learning.

### Hiện tượng Quá khớp (Overfitting)
Khi chúng ta huấn luyện mô hình trên một tập dữ liệu hữu hạn, có nguy cơ mô hình sẽ "học vẹt" quá kỹ các chi tiết nhiễu của dữ liệu huấn luyện mà mất đi khả năng dự đoán chính xác trên dữ liệu thực tế mới. Hiện tượng này gọi là **Quá khớp (Overfitting)**, và các kỹ thuật để chống lại nó được gọi là **Kiểm soát/Điều hòa (Regularization)**.

---

## 1. Sai số Huấn luyện (Training Error) và Sai số Khái quát hóa (Generalization Error)

Trong học máy giám sát, chúng ta thường dựa vào **Giả định IID**: giả định rằng dữ liệu huấn luyện và dữ liệu kiểm tra được rút ra một cách *độc lập* (independently) và từ cùng một *phân phối giống hệt nhau* (identical distributions).

Chúng ta cần phân biệt rõ hai loại sai số:

1. **Sai số Huấn luyện ($R_\textrm{emp}$)**: Là sai số tính toán trực tiếp trên tập dữ liệu huấn luyện hiện có (một chỉ số thống kê cụ thể):
   $$R_\textrm{emp}[\mathbf{X}, \mathbf{y}, f] = \frac{1}{n} \sum_{i=1}^n l(\mathbf{x}^{(i)}, y^{(i)}, f(\mathbf{x}^{(i)}))$$

2. **Sai số Khái quát hóa ($R$)**: Là sai số kỳ vọng trên toàn bộ phân phối dữ liệu gốc (bao gồm cả những dữ liệu tương lai chưa nhìn thấy):
   $$R[p, f] = E_{(\mathbf{x}, y) \sim P} [l(\mathbf{x}, y, f(\mathbf{x}))]$$

Vì không thể biết trước phân phối thực tế của toàn bộ thế giới, chúng ta không thể tính trực tiếp $R$. Thay vào đó, ta **ước lượng** sai số khái quát hóa bằng cách đánh giá mô hình trên một **tập kiểm tra (test set)** độc lập được giữ lại từ đầu.

### Độ phức tạp của Mô hình (Model Complexity)
* **Mô hình quá đơn giản + Dữ liệu nhiều**: Sai số huấn luyện và sai số khái quát hóa sẽ gần nhau, nhưng cả hai đều có thể rất cao (Mô hình không đủ sức học).
* **Mô hình quá phức tạp + Dữ liệu ít**: Sai số huấn luyện cực kỳ thấp (thậm chí bằng 0), nhưng sai số khái quát hóa sẽ rất cao (Mô hình học vẹt).

---

## 2. Quá khớp (Overfitting) hay Non khớp (Underfitting)?

* **Underfitting (Non khớp)**: Xảy ra khi cả sai số huấn luyện và sai số xác thực (validation error) đều cao. Lý do thường là mô hình quá đơn giản để nắm bắt quy luật của dữ liệu.
* **Overfitting (Quá khớp)**: Xảy ra khi sai số huấn luyện rất thấp nhưng sai số xác thực lại rất cao. Mô hình đã học cả những nhiễu (noise) của tập huấn luyện.

### Ví dụ: Khớp đường cong đa thức (Polynomial Curve Fitting)
Giả sử ta dùng đa thức bậc $d$ để dự đoán $y$ từ $x$:
$$\hat{y}= \sum_{i=0}^d x^i w_i$$
* Đa thức bậc càng cao (độ phức tạp lớn), mô hình càng dễ khớp hoàn hảo với tập huấn luyện (sai số huấn luyện bằng 0), nhưng đồ thị sẽ uốn lượn kỳ dị và dự đoán sai lệch hoàn toàn ở các điểm mới (Overfitting).

### Kích thước dữ liệu (Dataset Size)
* Dữ liệu càng ít, mô hình càng dễ bị quá khớp.
* Thêm nhiều dữ liệu chất lượng luôn luôn giúp cải thiện khả năng khái quát hóa và giảm quá khớp.

## 3. Lựa chọn Mô hình (Model Selection)

Để chọn ra mô hình tốt nhất (chọn kiến trúc, siêu tham số - hyperparameters), chúng ta **không bao giờ được phép dùng tập kiểm tra (test set)** để tinh chỉnh, vì làm vậy sẽ khiến mô hình bị quá khớp một cách gián tiếp lên tập kiểm tra.

Giải pháp là chia dữ liệu làm 3 phần:
1. **Tập Huấn luyện (Training set)**: Dùng để tối ưu hóa trọng số mô hình.
2. **Tập Xác thực (Validation set)**: Dùng để đánh giá và lựa chọn siêu tham số/mô hình tốt nhất.
3. **Tập Kiểm tra (Test set)**: Chỉ dùng duy nhất một lần cuối cùng để đánh giá khách quan sức mạnh thực tế của mô hình.

### Kiểm tra chéo K-Fold (K-Fold Cross-Validation)
Khi dữ liệu quá khan hiếm không đủ để chia riêng tập xác thực, ta dùng **K-fold cross-validation**:
* Chia dữ liệu huấn luyện thành $K$ phần không trùng lặp.
* Huấn luyện mô hình $K$ lần, mỗi lần dùng $K-1$ phần để huấn luyện và phần còn lại để xác thực.
* Sai số cuối cùng là trung bình cộng sai số của $K$ lần chạy.

---

## Tóm tắt quy tắc cốt lõi
1. Luôn dùng **tập xác thực** (hoặc K-fold cross-validation) để lựa chọn mô hình.
2. Mô hình càng phức tạp thì càng cần nhiều dữ liệu huấn luyện.
3. Nhiều dữ liệu hơn gần như luôn mang lại khả năng khái quát hóa tốt hơn.
4. Toàn bộ lý thuyết khái quát hóa này dựa trên **giả định IID** (dữ liệu độc lập và cùng phân phối).
