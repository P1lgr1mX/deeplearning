# Khái quát hóa trong Học sâu

Trong Chương hồi quy và phân loại, chúng ta đã giải quyết các bài toán hồi quy và phân loại bằng cách khớp các mô hình tuyến tính vào dữ liệu huấn luyện. Trong cả hai trường hợp, chúng ta đã cung cấp các thuật toán thực tế để tìm các tham số nhằm tối đa hóa khả năng xảy ra của các nhãn huấn luyện được quan sát. Và sau đó, về cuối mỗi chương, chúng ta đã nhớ lại rằng việc khớp dữ liệu huấn luyện chỉ là một mục tiêu trung gian. Nhiệm vụ thực sự của chúng ta từ đầu đến cuối là khám phá *các quy luật tổng quát* dựa trên đó chúng ta có thể đưa ra các dự đoán chính xác ngay cả trên các ví dụ mới được rút ra từ cùng một quần thể cơ sở. Các nhà nghiên cứu học máy là *người tiêu dùng* của các thuật toán tối ưu hóa. Đôi khi, chúng ta thậm chí phải phát triển các thuật toán tối ưu hóa mới. Nhưng cuối cùng, tối ưu hóa chỉ đơn thuần là phương tiện để đạt được mục đích. Về cốt lõi, học máy là một ngành thống kê và chúng ta chỉ mong muốn tối ưu hóa tổn thất huấn luyện trong chừng mực một số nguyên lý thống kê (đã biết hoặc chưa biết) dẫn dắt các mô hình kết quả khái quát hóa tốt hơn ngoài tập huấn luyện.

Tích cực mà nói, hóa ra các mạng thần kinh sâu được huấn luyện bằng phương pháp giảm cực độ gradient ngẫu nhiên khái quát hóa cực kỳ tốt trên vô số bài toán dự đoán, bao gồm thị giác máy tính, xử lý ngôn ngữ tự nhiên, dữ liệu chuỗi thời gian, hệ thống gợi ý, hồ sơ sức khỏe điện tử, gấp protein, xấp xỉ hàm giá trị trong trò chơi điện tử và trò chơi bàn cờ, cùng nhiều lĩnh vực khác. Mặt tiêu cực là, nếu bạn đang tìm kiếm một lời giải thích đơn giản về câu chuyện tối ưu hóa (tại sao chúng ta có thể khớp chúng với dữ liệu huấn luyện) hoặc câu chuyện khái quát hóa (tại sao các mô hình kết quả lại khái quát hóa tốt cho các ví dụ chưa từng thấy), thì bạn có thể muốn rót cho mình một ly nước để bình tĩnh lại. Trong khi các quy trình tối ưu hóa mô hình tuyến tính và các thuộc tính thống kê của các giải pháp đều được mô tả tốt bởi một hệ thống lý thuyết toàn diện, thì hiểu biết của chúng ta về học sâu vẫn giống như vùng "miền Tây hoang dã" trên cả hai mặt trận.

Cả lý thuyết và thực hành học sâu đều đang phát triển nhanh chóng, với các nhà lý thuyết áp dụng các chiến lược mới để giải thích những gì đang diễn ra, ngay cả khi các nhà thực hành tiếp tục đổi mới với tốc độ chóng mặt, xây dựng kho vũ khí gồm các thuật toán heuristic để huấn luyện mạng sâu và một hệ thống trực giác cũng như kiến thức dân gian cung cấp hướng dẫn để quyết định kỹ thuật nào áp dụng trong tình huống nào.

Tóm tắt của thời điểm hiện tại là lý thuyết học sâu đã tạo ra các hướng tấn công đầy hứa hẹn và các kết quả hấp dẫn rải rác, nhưng dường như vẫn còn xa mới có một giải thích toàn diện về cả (i) tại sao chúng ta có thể tối ưu hóa mạng thần kinh và (ii) làm thế nào các mô hình học được bằng phương pháp giảm gradient lại có thể khái quát hóa tốt như vậy, ngay cả trên các tác vụ nhiều chiều. Tuy nhiên, trong thực tế, (i) hiếm khi là vấn đề (chúng ta luôn có thể tìm thấy các tham số phù hợp với tất cả dữ liệu huấn luyện của mình) và do đó hiểu được khái quát hóa là vấn đề lớn hơn nhiều. Mặt khác, ngay cả khi thiếu đi một lý thuyết khoa học mạch lạc, các nhà thực hành đã phát triển một bộ sưu tập lớn các kỹ thuật có thể giúp bạn tạo ra các mô hình khái quát hóa tốt trong thực tế. Mặc dù không có bản tóm tắt ngắn gọn nào có thể công bằng cho chủ đề khái quát hóa rộng lớn trong học sâu, và mặc dù trạng thái nghiên cứu tổng thể còn lâu mới được giải quyết, chúng tôi hy vọng, trong phần này, sẽ trình bày một cái nhìn tổng quan rộng rãi về trạng thái nghiên cứu và thực hành.

## Xem xét lại Quá khớp (Overfitting) và Điều chuẩn (Regularization)

Theo định lý "không có bữa ăn trưa nào miễn phí" của Wolpert (1995), bất kỳ thuật toán học máy nào cũng khái quát hóa tốt hơn trên dữ liệu có phân phối nhất định và kém hơn với các phân phối khác. Do đó, với một tập huấn luyện hữu hạn, một mô hình dựa trên các giả định nhất định: để đạt được hiệu suất ngang tầm con người, việc xác định *thiên kiến quy nạp* (inductive biases) phản ánh cách con người nghĩ về thế giới có thể hữu ích. Các thiên kiến quy nạp như vậy thể hiện sự ưu tiên cho các giải pháp có các thuộc tính nhất định. Ví dụ, một MLP sâu có thiên kiến quy nạp hướng tới việc xây dựng một hàm phức tạp bằng cách kết hợp các hàm đơn giản hơn.

Với các mô hình học máy mã hóa các thiên kiến quy nạp, phương pháp huấn luyện chúng điển hình gồm hai giai đoạn: (i) khớp dữ liệu huấn luyện; và (ii) ước lượng *sai số khái quát hóa* (sai số thực trên quần thể cơ sở) bằng cách đánh giá mô hình trên dữ liệu kiểm thử được giữ lại. Sự khác biệt giữa mức độ khớp trên dữ liệu huấn luyện và mức độ khớp trên dữ liệu kiểm thử được gọi là *khoảng cách khái quát hóa* (generalization gap) và khi khoảng cách này lớn, chúng ta nói rằng mô hình của chúng ta *quá khớp* với dữ liệu huấn luyện. Trong các trường hợp quá khớp cực đoan, chúng ta có thể khớp chính xác dữ liệu huấn luyện, ngay cả khi sai số kiểm thử vẫn ở mức đáng kể. Và theo quan điểm cổ điển, cách giải thích là mô hình của chúng ta quá phức tạp, đòi hỏi chúng ta phải giảm bớt số lượng đặc trưng, số lượng tham số học được khác không, hoặc kích thước của các tham số như đã định lượng.

Tuy nhiên, học sâu làm phức tạp bức tranh này theo những cách trái ngược với trực giác. Đầu tiên, đối với các bài toán phân loại, các mô hình của chúng ta thường đủ mạnh để khớp hoàn hảo từng ví dụ huấn luyện, ngay cả trong các bộ dữ liệu bao gồm hàng triệu mẫu (Zhang và cộng sự, 2021). Trong bức tranh cổ điển, chúng ta có thể nghĩ rằng thiết lập này nằm ở cực bên phải của trục độ phức tạp của mô hình, và bất kỳ sự cải thiện nào về sai số khái quát hóa phải đến từ việc điều chuẩn, bằng cách giảm độ phức tạp của lớp mô hình hoặc bằng cách áp dụng một hình phạt, hạn chế nghiêm ngặt tập hợp các giá trị mà các tham số của chúng ta có thể nhận. Nhưng đó là nơi mọi thứ bắt đầu trở nên kỳ lạ.

Kỳ lạ thay, đối với nhiều tác vụ học sâu (ví dụ: nhận dạng hình ảnh và phân loại văn bản), chúng ta thường chọn các kiến trúc mô hình mà tất cả chúng đều có thể đạt được tổn thất huấn luyện cực kỳ thấp (và sai số huấn luyện bằng không). Bởi vì tất cả các mô hình đang được xem xét đều đạt được sai số huấn luyện bằng không, *con đường duy nhất để đạt được những tiến bộ xa hơn là giảm quá khớp*. Thậm chí kỳ lạ hơn, thường là trường hợp mặc dù khớp hoàn hảo dữ liệu huấn luyện, chúng ta thực sự có thể *giảm sai số khái quát hóa* sâu hơn nữa bằng cách làm cho mô hình *thậm chí còn mạnh hơn*, ví dụ: thêm các lớp, các nút hoặc huấn luyện với số lượng epoch lớn hơn. Kỳ lạ hơn nữa, quy luật liên quan giữa khoảng cách khái quát hóa và *độ phức tạp* của mô hình (được thể hiện qua độ sâu hoặc độ rộng của mạng) có thể không đơn điệu, với độ phức tạp lớn hơn sẽ gây hại lúc đầu nhưng sau đó lại giúp ích trong một mô hình gọi là "double-descent" (Nakkiran và cộng sự, 2021). Do đó, nhà thực hành học sâu sở hữu một túi mẹo, một số mẹo dường như hạn chế mô hình theo một cách nào đó và những mẹo khác dường như làm cho nó trở nên mạnh mẽ hơn, và tất cả chúng, theo một nghĩa nào đó, đều được áp dụng để giảm thiểu quá khớp.

Làm cho mọi thứ trở nên phức tạp hơn nữa, trong khi các đảm bảo được cung cấp bởi lý thuyết học cổ điển có thể mang tính bảo thủ ngay cả đối với các mô hình cổ điển, chúng dường như bất lực trong việc giải thích tại sao các mạng thần kinh sâu lại có thể khái quát hóa tốt ngay từ đầu. Bởi vì các mạng thần kinh sâu có khả năng khớp các nhãn tùy ý ngay cả đối với các bộ dữ liệu lớn, và bất chấp việc sử dụng các phương pháp quen thuộc như điều chuẩn $\ell_2$, các giới hạn khái quát hóa dựa trên độ phức tạp truyền thống (ví dụ: những giới hạn dựa trên chiều VC hoặc độ phức tạp Rademacher của một lớp giả thuyết) không thể giải thích tại sao mạng thần kinh lại khái quát hóa được.

## Cảm hứng từ Phương pháp Phi tham số (Nonparametrics)

Tiếp cận học sâu lần đầu tiên, thật dễ bị cám dỗ để nghĩ về chúng như các mô hình tham số. Sau cùng, các mô hình *thực sự* có hàng triệu tham số. Khi chúng ta cập nhật các mô hình, chúng ta cập nhật các tham số của chúng. Khi chúng ta lưu các mô hình, chúng ta ghi các tham số của chúng vào đĩa. Tuy nhiên, toán học và khoa học máy tính chứa đầy những thay đổi quan điểm trái ngược với trực giác và các đẳng cấu đáng ngạc nhiên giữa các bài toán tưởng chừng như khác nhau. Mặc dù các mạng thần kinh rõ ràng *có* tham số, nhưng trong một số cách, việc nghĩ về chúng hoạt động giống như các mô hình phi tham số có thể mang lại nhiều kết quả hơn. Vậy chính xác điều gì làm cho một mô hình trở thành phi tham số? Mặc dù cái tên bao gồm một tập hợp các cách tiếp cận đa dạng, một chủ đề chung là các phương pháp phi tham số có xu hướng có mức độ phức tạp tăng lên khi lượng dữ liệu có sẵn tăng lên.

Có lẽ ví dụ đơn giản nhất về mô hình phi tham số là thuật toán $k$-láng giềng gần nhất (k-nearest neighbor). Ở đây, tại thời điểm huấn luyện, người học chỉ đơn giản là ghi nhớ bộ dữ liệu. Sau đó, tại thời điểm dự đoán, khi đối mặt với một điểm mới $\mathbf{x}$, người học sẽ tra cứu $k$ láng giềng gần nhất (tức $k$ điểm $\mathbf{x}_i'$ tối thiểu hóa khoảng cách $d(\mathbf{x}, \mathbf{x}_i')$). Khi $k=1$, thuật toán này được gọi là $1$-láng giềng gần nhất, và thuật toán sẽ luôn đạt được sai số huấn luyện bằng không. Tuy nhiên, điều đó không có nghĩa là thuật toán sẽ không khái quát hóa được. Trên thực tế, hóa ra trong một số điều kiện ôn hòa, thuật toán 1-láng giềng gần nhất là nhất quán (cuối cùng hội tụ về bộ dự đoán tối ưu).

Lưu ý rằng $1$-láng giềng gần nhất yêu cầu chúng ta chỉ định một số hàm khoảng cách $d$, hoặc tương đương, chúng ta chỉ định một số hàm cơ sở giá trị vector $\phi(\mathbf{x})$ để biểu diễn dữ liệu của chúng ta. Đối với bất kỳ lựa chọn nào về số đo khoảng cách, chúng ta sẽ đạt được sai số huấn luyện bằng không và cuối cùng đạt được một bộ dự đoán tối ưu, nhưng các số đo khoảng cách khác nhau $d$ mã hóa các thiên kiến quy nạp khác nhau và với một lượng dữ liệu hữu hạn có sẵn sẽ mang lại các bộ dự đoán khác nhau. Các lựa chọn khác nhau của số đo khoảng cách $d$ đại diện cho các giả định khác nhau về các quy luật cơ sở và hiệu suất của các bộ dự đoán khác nhau sẽ phụ thuộc vào mức độ tương thích của các giả định đó với dữ liệu quan sát được.

Theo một nghĩa nào đó, bởi vì các mạng thần kinh bị quá tham số hóa, sở hữu nhiều tham số hơn mức cần thiết để khớp dữ liệu huấn luyện, chúng có xu hướng *nội suy* dữ liệu huấn luyện (khớp nó một cách hoàn hảo) và do đó hành xử, theo một số cách, giống như các mô hình phi tham số hơn. Nghiên cứu lý thuyết gần đây đã thiết lập mối liên hệ sâu sắc giữa các mạng thần kinh lớn và các phương pháp phi tham số, đặc biệt là các phương pháp hạt nhân (kernel methods). Cụ thể, Jacot và cộng sự (2018) đã chứng minh rằng trong giới hạn, khi các perceptron đa lớp với các trọng số được khởi tạo ngẫu nhiên phát triển rộng vô hạn, chúng trở nên tương đương với các phương pháp hạt nhân (phi tham số) cho một lựa chọn cụ thể của hàm hạt nhân (về cơ bản là một hàm khoảng cách), mà họ gọi là hạt nhân tiếp tuyến thần kinh (neural tangent kernel). Mặc dù các mô hình hạt nhân tiếp tuyến thần kinh hiện tại có thể không giải thích đầy đủ hành vi của các mạng sâu hiện đại, nhưng thành công của chúng như một công cụ phân tích nhấn mạnh tính hữu ích của mô hình phi tham số để hiểu hành vi của các mạng sâu quá tham số hóa.

## Dừng sớm (Early Stopping)

Mặc dù các mạng thần kinh sâu có khả năng khớp các nhãn tùy ý, ngay cả khi các nhãn được gán sai hoặc ngẫu nhiên (Zhang và cộng sự, 2021), khả năng này chỉ xuất hiện qua nhiều vòng lặp huấn luyện. Một hướng nghiên cứu mới (Rolnick và cộng sự, 2017) đã tiết lộ rằng trong thiết lập nhiễu nhãn, các mạng thần kinh có xu hướng khớp dữ liệu được dán nhãn sạch trước và chỉ sau đó mới nội suy dữ liệu bị dán nhãn sai. Hơn nữa, người ta đã thiết lập rằng hiện tượng này dịch trực tiếp thành một đảm bảo về khả năng khái quát hóa: bất khi nào một mô hình khớp với dữ liệu được dán nhãn sạch nhưng chưa khớp các ví dụ được dán nhãn ngẫu nhiên có trong tập huấn luyện, nó thực tế đã khái quát hóa (Garg và cộng sự, 2021).

Cùng với nhau, những phát hiện này giúp thúc đẩy phương pháp *dừng sớm* (early stopping), một kỹ thuật cổ điển để điều chuẩn các mạng thần kinh sâu. Ở đây, thay vì trực tiếp hạn chế các giá trị của trọng số, người ta hạn chế số lượng epoch huấn luyện. Cách phổ biến nhất để xác định tiêu chí dừng là theo dõi sai số kiểm định trong suốt quá trình huấn luyện (thường bằng cách kiểm tra một lần sau mỗi epoch) và cắt bỏ huấn luyện khi sai số kiểm định không giảm quá một lượng nhỏ $\epsilon$ trong một số epoch nhất định. Điều này đôi khi được gọi là *tiêu chí kiên nhẫn* (patience criterion). Ngoài khả năng mang lại sự khái quát hóa tốt hơn trong bối cảnh nhãn bị nhiễu, một lợi ích khác của việc dừng sớm là tiết kiệm thời gian. Khi tiêu chí kiên nhẫn được đáp ứng, người ta có thể chấm dứt huấn luyện. Đối với các mô hình lớn có thể yêu cầu nhiều ngày huấn luyện đồng thời trên tám hoặc nhiều GPU hơn, việc dừng sớm được tinh chỉnh tốt có thể tiết kiệm cho các nhà nghiên cứu nhiều ngày thời gian và có thể tiết kiệm cho tổ chức của họ hàng ngàn đô la.

Đáng chú ý, khi không có nhiễu nhãn và các bộ dữ liệu là khả thi (các lớp thực sự có thể phân tách được, ví dụ: phân biệt mèo với chó), dừng sớm có xu hướng không dẫn đến những cải thiện đáng kể trong khả năng khái quát hóa. Mặt khác, khi có nhiễu nhãn, hoặc có sự biến đổi nội tại trong nhãn (ví dụ: dự đoán tỷ lệ tử vong ở bệnh nhân), dừng sớm là cực kỳ quan trọng. Huấn luyện các mô hình cho đến khi chúng nội suy dữ liệu nhiễu thường là một ý tưởng tồi.

## Các phương pháp điều chuẩn cổ điển cho mạng sâu

Trong Chương hồi quy, chúng tôi đã mô tả một số kỹ thuật điều chuẩn cổ điển để hạn chế độ phức tạp của các mô hình của chúng ta. Cụ thể, phần suy giảm trọng số đã giới thiệu một phương pháp gọi là suy giảm trọng số (weight decay), bao gồm việc thêm một thuật ngữ điều chuẩn vào hàm tổn thất để phạt các giá trị lớn của trọng số. Tùy thuộc vào chuẩn trọng số nào bị phạt, kỹ thuật này được gọi là điều chuẩn Ridge (cho hình phạt $\ell_2$) hoặc điều chuẩn Lasso (cho hình phạt $\ell_1$). Trong phân tích cổ điển về các bộ điều chuẩn này, chúng được coi là đủ hạn chế đối với các giá trị mà trọng số có thể nhận để ngăn mô hình khớp với các nhãn tùy ý.

Trong các triển khai học sâu, suy giảm trọng số vẫn là một công cụ phổ biến. Tuy nhiên, các nhà nghiên cứu đã lưu ý rằng sức mạnh điển hình của điều chuẩn $\ell_2$ là không đủ để ngăn các mạng nội suy dữ liệu (Zhang và cộng sự, 2021) và do đó, các lợi ích nếu được giải thích dưới dạng điều chuẩn có thể chỉ có ý nghĩa khi kết hợp với tiêu chí dừng sớm. Nếu không có dừng sớm, có khả năng giống như số lượng lớp hoặc số lượng nút (trong học sâu) hoặc số đo khoảng cách (trong 1-láng giềng gần nhất), các phương pháp này có thể dẫn đến khái quát hóa tốt hơn không phải vì chúng hạn chế một cách có ý nghĩa sức mạnh của mạng thần kinh mà là vì chúng bằng cách nào đó mã hóa các thiên kiến quy nạp tương thích tốt hơn với các quy luật được tìm thấy trong các bộ dữ liệu quan tâm. Do đó, các bộ điều chuẩn cổ điển vẫn phổ biến trong các triển khai học sâu, ngay cả khi lý do lý thuyết cho hiệu quả của chúng có thể hoàn toàn khác nhau.

Đáng chú ý, các nhà nghiên cứu học sâu cũng đã xây dựng dựa trên các kỹ thuật lần đầu tiên được phổ biến trong các bối cảnh điều chuẩn cổ điển, chẳng hạn như thêm nhiễu vào các đầu vào của mô hình. Trong phần tiếp theo, chúng tôi sẽ giới thiệu kỹ thuật dropout nổi tiếng (được phát minh bởi Srivastava và cộng sự, 2014), kỹ thuật này đã trở thành trụ cột của học sâu, ngay cả khi cơ sở lý thuyết cho hiệu quả của nó vẫn tương tự như vậy là một điều bí ẩn.

## Tóm tắt

Không giống như các mô hình tuyến tính cổ điển, thường có ít tham số hơn các ví dụ, các mạng sâu có xu hướng bị quá tham số hóa và đối với hầu hết các tác vụ đều có khả năng khớp hoàn hảo với tập huấn luyện. *Chế độ nội suy* này thách thức nhiều trực giác lâu đời. Về mặt chức năng, các mạng thần kinh trông giống như các mô hình tham số. Nhưng nghĩ về chúng như các mô hình phi tham số đôi khi có thể là một nguồn trực giác đáng tin cậy hơn. Bởi vì thường xảy ra trường hợp tất cả các mạng sâu đang được xem xét đều có khả năng khớp với tất cả các nhãn huấn luyện, hầu hết tất cả các lợi ích phải đến từ việc giảm thiểu quá khớp (thu hẹp *khoảng cách khái quát hóa*). Trớ trêu thay, các can thiệp làm giảm khoảng cách khái quát hóa đôi khi dường như làm tăng độ phức tạp của mô hình và vào những thời điểm khác lại dường như làm giảm độ phức tạp. Tuy nhiên, các phương pháp này hiếm khi làm giảm độ phức tạp đủ để lý thuyết cổ điển giải thích sự khái quát hóa của các mạng sâu, và *tại sao một số lựa chọn nhất định lại dẫn đến khả năng khái quát hóa được cải thiện* phần lớn vẫn là một câu hỏi mở lớn bất chấp những nỗ lực phối hợp của nhiều nhà nghiên cứu xuất sắc.
